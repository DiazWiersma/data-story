# Introductie

Sinds de coronapandemie is thuiswerken in een stroomversnelling geraakt. In Nederland werkt inmiddels meer dan de helft van de beroepsbevolking, zo’n vijf miljoen mensen, weleens vanuit huis (CBS, 2024). Maar wat betekent dit voor werknemers? Over de effecten van thuiswerken lopen de meningen uiteen. Voorstanders wijzen op hogere productiviteit, minder stress en een betere werk-privébalans. Tegenstanders waarschuwen juist voor sociale isolatie en afleidingen in de thuissituatie. Naast het thuiswerken is er ook discussie over de lengte van de werkweek. In Nederland ligt het gemiddelde op 32 uur, maar is meer werken altijd beter? Of heeft een lange werkweek juist negatieve gevolgen, zoals hogere stressniveaus en verminderde welzijn? In dit data story project worden twee perspectieven onderzocht: Thuiswerken verhoogt de productiviteit van werknemers. Een toename in het aantal gewerkte uren heeft negatieve effecten op de werknemer. Aan de hand van zelfgerapporteerde data uit remote-work-enquêtes, met metingen van onder andere productiviteit, stress, focus en gevoelens van isolatie, worden deze perspectieven onderbouwd en gevisualiseerd. Zo ontstaat een genuanceerd beeld van de impact van werkvormen en werkdruk op werknemers.
