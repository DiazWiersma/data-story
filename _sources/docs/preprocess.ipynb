# Datasets en Preprocessing
Om onze perspectieven te vormen voor deze opdracht, begonnen we met een korte brainstorm over mogelijke thema’s. Al snel werd gekozen voor het onderwerp *remote werken en productiviteit*. Hoewel de gevonden datasets interessante inzichten boden op het gebied van productiviteit en welzijn, bleek het lastig om een duidelijk tweede perspectief te formuleren, er ontbrak bijvoorbeeld een uitgesproken negatieve kant van remote werken, of een sterk afwijkend positief effect. De resultaten vielen vaak ergens in het midden.

Daarom kozen we ervoor om ons tweede perspectief te richten op een ander relevant aspect van modern werken: het aantal gewerkte uren. Uiteindelijk formuleerden we de volgende twee perspectieven:

1. *Remote werken leidt tot productievere en gelukkigere werknemers*

2. *Minder werkuren leiden tot betere uitkomsten dan meer werkuren*

Op basis hiervan is definitief gekozen voor de volgende vier datasets.

## Dataset beschrijvingen
### Dataset 1: remote_work_productivity
Deze dataset bevat zelfgerapporteerde informatie over werknemers die op afstand werken of op kantoor. De focus ligt op productiviteit en welzijn, met variabelen zoals Productivity_Score, Well_Being_Score, Employment_Type en Hours_Worked_Per_Week. Het doel is om de impact van de werklocatie op productiviteit en welzijn te onderzoeken.

*https://www.kaggle.com/datasets/mrsimple07/remote-work-productivity*

### Dataset 2: The Impacts of Working Remotely and in an Office Survey
Deze survey-gebaseerde dataset onderzoekt de perceptie van werknemers over werken op afstand versus op kantoor. De gegevens bevatten meningen over productiviteit, communicatie, motivatie en werktevredenheid, en bieden inzicht in hoe werknemers hun werkomgeving ervaren.

*https://www.kaggle.com/datasets/mohamedelzeini/the-impacts-of-working-remotely-and-in-an-office?select=The+Impacts+of+Working+Remotely+and+in+an+Office+Survey.csv*

### Dataset 3: Impact_of_Remote_Work_on_Mental_Health
Deze dataset richt zich op het verband tussen thuiswerken en mentale gezondheid. Respondenten geven informatie over stressniveaus, mentale aandoeningen en slaapkwaliteit, ingedeeld per werklocatie (zoals "Home" of "Office"). Dit maakt de dataset geschikt voor welzijnsanalyses binnen verschillende werkcontexten.

*https://www.kaggle.com/datasets/waqi786/remote-work-and-mental-health*

### Dataset 4: Extended_Employee_Performance_and_Productivity_Data
Deze uitgebreide dataset bevat informatie over werknemersprestaties, zoals maandelijkse salarissen, werkuren, afdelingen en functietitels (bijv. "Consultant"). De dataset is nuttig voor het analyseren van verbanden tussen werkbelasting en compensatie binnen verschillende bedrijfsfuncties.

*https://www.kaggle.com/datasets/mexwell/employee-performance-and-productivity-data*


## Preprocessing
### Cleaning

Voor deze analyse zijn vier datasets ingeladen, elk met een andere focus binnen het thema **remote work** (productiviteit, welzijn, mentale gezondheid en salarissen). De preprocessing en cleaning zijn per dataset uitgevoerd, afhankelijk van de structuur en kwaliteit van de data.

#### Algemene schoonmaakacties:

* **Verwijderen van overbodige spaties** in kolomnamen en waardes (bijv. `str.strip()`).
* **Standaardisatie van categorische waarden**, zoals het kapitaliseren van waarden in kolommen als `Work_Location`, `Stress_Level` en `Sleep_Quality`.
* **Omgaan met ontbrekende waarden**:

  * In de `Mental_Health_Condition`-kolom zijn lege waarden vervangen door `"None"` om analyse mogelijk te maken.
* **Binning van continue variabelen**:

  * In de consultant dataset zijn werkuren gebundeld in categorieën van 5 uur (`30–34`, `35–39`, etc.) via `pd.cut()` om trends in salarisontwikkeling per werkuren inzichtelijk te maken.

#### Data filtering:

* Alleen respondenten met specifieke kenmerken zijn geselecteerd in bepaalde analyses. Zo zijn in één analyse (figure 8) alleen **consultants** meegenomen en uitgesplitst naar **afdeling** en **werkuren**.

---

### Variabelen

De variabelen die in de analyses worden gebruikt, zijn een mix van continue en categorische variabelen. Hieronder volgt een classificatie per type meetschaal:

#### Continuous / Ratio:

* `Productivity_Score`
* `Well_Being_Score`
* `Hours_Worked_Per_Week`
* `Monthly_Salary`

#### Categorical / Nominal:

* `Employment_Type` (Remote / On-Site)
* `Work_Location` (Home, Office, etc.)
* `Mental_Health_Condition`
* `Stress_Level`
* `Sleep_Quality`
* `Department`
* `Job_Title`

#### Ordinal:

* `Hours_Bin` (gecategoriseerde werkuren per week)

---

### Aggregaties

Verschillende vormen van aggregatie zijn toegepast om inzichten te verkrijgen:

* **Gemiddelde productiviteit en welzijn per werkvorm**:

  * Groepering op `Employment_Type`, gevolgd door `.mean()` op `Productivity_Score` en `Well_Being_Score`.

* **Percentage berekeningen voor welzijnsindicatoren**:

  * Voor elke `Work_Location` is het percentage respondenten berekend dat:

    * een **lage stress** ervaart,
    * **geen mentale aandoeningen** heeft,
    * **goede slaapkwaliteit** rapporteert.
  * De uitkomsten zijn uitgedrukt als percentages van het totaal per locatie.

* **Binning en gemiddelden van salarissen per uurcategorie**:

  * Binnen de consultantgroep is het gemiddelde **maandsalaris** per combinatie van `Department` en `Hours_Bin` berekend.
  * Dit maakt trends zichtbaar in salaris versus werkbelasting, uitgesplitst per afdeling.
