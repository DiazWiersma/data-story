# Reflectie
## Peer feedback
Op 24 juni 2025 verzamelde onze groep F1 zich samen met groep F2 voor een peer feedbacksessie. Onze groep presenteerde als eerste de resultaten van drie weken werk, waarna we feedback ontvingen van groep F2. Vervolgens gaven wij zelf feedback op het werk van groep F2.

De algemene indruk van ons data story-project was positief. Er werden echter enkele aandachtspunten genoemd door zowel het andere team als de TA. Zo werd benadrukt dat er meer variatie in de grafieken nodig is, dus meer verschillende soorten visualisaties. Daarnaast werd opgemerkt dat er een regressielijn in een scatterplot kon worden toegevoegd. Tot slot moest de introductie en de argumenten meer uitgewerkt worden.

## Persoonlijke reflectie
Een belangrijk leerpunt was dat onze oorspronkelijke planning te ambitieus was. We merkten bij het maken van de eerste versie dat we niet genoeg tijd hadden ingepland om geschikte datasets te zoeken en te analyseren. De datasets die we in eerste instantie hadden verzameld, sloten niet goed aan op onze gekozen perspectieven en argumenten. Daardoor moesten we van onderwerp veranderen en nieuwe datasets verzamelen, wat ons uiteindelijk minder tijd gaf om de data story rond het nieuwe onderwerp uit te werken.

Daarnaast realiseerden we ons dat onze visualisaties soms nuance misten. Hoewel de figuren en grafieken duidelijk de belangrijkste trends en verschillen lieten zien, zoals het verschil in zelfgerapporteerde productiviteit tussen remote en in-office werknemers of de negatieve relatie tussen werkuren en welzijn en productiviteit, toonden de visualisaties vaak alleen gemiddelden of trends. Spreiding, foutmarges of significantie-indicatoren werden niet altijd weergegeven. Hierdoor werd de complexiteit van de data mogelijk niet volledig zichtbaar en kon de interpretatie te simplistisch overkomen. Ook waren de meeste visualisaties gebaseerd op subjectieve percepties, zoals zelfgerapporteerde productiviteit en welzijn. Dit was visueel overtuigend, maar het kan bij de lezer de indruk wekken dat het om objectieve uitkomsten gaat.
