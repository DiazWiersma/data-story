## De kracht van remote werken

### Hogere zelfgerapporteerde productiviteit

Deze visualisatie laat zien dat werknemers zichzelf over het algemeen als productiever ervaren in een remote werkomgeving dan op een fysieke werkplek.

De in Figuur 1 getoonde verschillen in productiviteitsscore (gemiddeld 73 voor remote medewerkers tegenover 64 voor kantoormedewerkers) bevestigen deze perceptie deels, maar het verschil blijft beperkt. De data zeggen bovendien niets over objectieve prestaties output of meetbare resultaten. 

Een mogelijke verklaring voor de hogere scores onder remote werkenden kan liggen in factoren als minder afleiding door collega's, meer flexibiliteit of het wegvallen van reistijd. Tegelijkertijd kunnen persoonlijke voorkeuren, thuissituatie of het type functie ook van invloed zijn op deze ervaring. In de huidige analyse is geen onderscheid gemaakt tussen functietypen of sectoren, terwijl die mogelijk sterk verschillen in de mate waarin remote werken als productief wordt ervaren. Zo kan een softwareontwikkelaar gebaat zijn bij geconcentreerd thuiswerken zonder afleiding, terwijl een salesmedewerker juist profiteert van directe interactie op kantoor met collega’s of klanten.

### Minder stress en betere wellbeing

Remote workers geven aan productiever te zijn, maar heeft remote working ook invloed op hun mentale gezondheid?

Uit de gegevens van een survey binnen een bedrijf waar aan zowel onsite, remote als hybrid (soms onsite en soms remote) werknemers werd gevraagd of zij stress ervaarden, een mentale aandoening hadden en of ze goed slapen blijkt dat de aantallen werknemers die aangaven dat zij lage stress ervaarden, geen mentale aandoening hadden en goed slapen ongeveer gelijk zijn van zowel onsite werknemers, remote werknemers als hybrid  werknemers.

```python
HTML('''<div style="text-align: left;"><img src="../_static/Graphs/figure_4.png" width="1000"></div>''')
```

> *Figuur 3: Vergelijking van welzijnsindicatoren tussen remote, onsite en hybride werkers*

Uit de dataset van de survey waar werknemers werd gevraagd hoe zij hun productiviteit ervaarden werd hen ook gevraagd om zichzelf een welzijnsscore te geven tussen 1 en 100. In figuur 4 is te zien dat remote werknemers zichzelf gemiddeld een hogere score geven dan in-office werknemers. 

```python
HTML('''<div style="text-align: center;"><img src="../_static/Graphs/figure_5.png" width="700"></div>''')
```

Hoewel de welzijnsindicatoren nauwelijks verschillen tussen de werkvormen, valt op dat remote werknemers zichzelf subjectief een hoger welzijn toekennen. Dit verschil tussen gezondheid en de ervaren welzijn suggereert dat door bijvoorbeeld de autonomie, flexibiliteit en minder reistijd die remote werken met zich meebrengt een positieve invloed kan hebben op hoe werknemers hun werk en welzijn beleven, zonder dat dat misschien direct zichtbaar is in hun gezondheid.

> *Figuur 7: relatie tussen maandsalaris en het aantal werkuren per week*

Figuur 7 toont de relatie tussen het gemiddeld maandsalaris en het aantal gewerkte uren per week onder consultants, uitgesplitst per afdeling. Het valt op dat er geen eenduidig patroon zichtbaar is waarbij meer werken altijd leidt tot een hoger salaris. Sommige afdelingen zoals IT en HR laten wel een stijging in salaris zien bij toenemende werkuren vooral in de categorieën 45-49 en 55-59 uur per week. Voor andere afdelingen, zoals Operations en Finance, daalt het gemiddeld salaris juist bij een hoger aantal werkuren.

De verschillen tussen afdelingen zijn opmerkelijk klein: het gemiddelde salaris per groep varieert  tussen ongeveer €7.080 en €7.250. De HR-afdeling verdient over het algemeen het meeste, met een piek bij 55-59 uur per week. Voor Marketing en Engineering liggen de gemiddelde salarissen doorgaans aan de lagere kant, ongeacht hoeveel er gewerkt wordt.

Er is geen duidelijk verband zichtbaar tussen het aantal gewerkte uren per week en het gemiddeld maandsalaris. Het lijkt erop dat andere factoren, mogelijk afdeling, rol, prestatie en werkervaring  belangrijker zijn.
