# Reflectie
## Peer feedback

Op 24 juni 2025 verzamelde onze groep F1 zich samen met groep F2 voor een peer feedbacksessie. Onze groep presenteerde als eerste de resultaten van drie weken werk, waarna we feedback ontvingen van groep F2. Vervolgens gaven wij zelf feedback op het werk van groep F2.

De algemene indruk van ons data story-project was positief. Er werden echter enkele aandachtspunten genoemd door zowel het andere team als de TA. Zo werd benadrukt dat er meer variatie in de grafieken nodig is, dus meer verschillende soorten aan variaties. Daarnaast werd opgemerkt dat er een regressie lijn in een scatterplot kon worden gemaakt. Tot slot moest de introductie en de argumenten meer uitgewerkt worden.

## Persoonlijke reflectie

Een ander mogelijk verbeterpunt voor ons data-story project was de planning. We merkten dat bij het maken van de eerste draft dat onze oorspronkelijke planning te ambitieus was. We hadden niet genoeg tijd ingepland om geschikte datasets te zoeken en te analyseren waardoor de datasets die we eerst hadden verzameld niet goed aansloten op onze oorspronkelijk gekozen perspectieven en argumenten. Daardoor moesten we van onderwerp veranderen en nieuwe datasets verzamelen en dit gaf ons minder tijd om met het nieuwe onderwerp de data-story te maken.
